```python
%%writefile main.cpp
#include <opencv2/opencv.hpp>
#include <filesystem>
#include <iostream>

namespace fs = std::filesystem;

int main() {
    std::string inputDir = "/content/images";
    std::string outputDir = "/content/resized";

    fs::create_directories(outputDir);

    for (const auto& file : fs::directory_iterator(inputDir)) {
        if (!file.is_regular_file()) continue;

        cv::Mat img = cv::imread(file.path().string());
        if (img.empty()) continue;

        cv::Mat resized;
        cv::resize(img, resized, cv::Size(1024, 768));

        std::string output =
            outputDir + "/" + file.path().filename().string();

        cv::imwrite(output, resized);
        std::cout << "Kaydedildi: " << output << std::endl;
    }

    return 0;
}
```
